```python
from Bio import SeqIO
records = list(SeqIO.parse("human_kinase_real.fasta", "fasta"))
print("Number of protein:", len(records))
print(records[0].id)
```

```python
from Bio.SeqUtils.ProtParam import ProteinAnalysis
def analyze_protein(seq):
    analyze = ProteinAnalysis(seq)
    aa_percent = analyze.get_amino_acids_percent()
    return{"MW": analyze.molecular_weight(), "PI": analyze.isoelectric_point(), 
    "A%": aa_percent["A"],
    "R%": aa_percent["R"],
    "N%": aa_percent["N"],
    "D%": aa_percent["D"],
    "K%": aa_percent["K"],
    "E%": aa_percent["E"]}
```

```python
import pandas as pd 
results = []
for record in records:
    seq = str(record.seq)
    stats = analyze_protein(seq)
    stats["protein_ID"] = record.id
    stats["length"] = len(seq)
    results.append(stats)
df = pd.DataFrame(results)
print(df)
```

```python
df.to_csv("protein_summary.csv", index=False)
print("file sucessfully save")
```

```python
from Bio.Blast import NCBIWWW, NCBIXML

test_seq = str(records[0].seq)
result_handle = NCBIWWW.qblast("blastp", "swissprot", test_seq)

blast_record = NCBIXML.read(result_handle)
top_hit = blast_record.alignments[0]
print("top hit:", top_hit.title)
print("E-value:", blast_record.alignments[0].hsps[0].expect)
print("score:", blast_record.alignments[0].hsps[0].score)
```
